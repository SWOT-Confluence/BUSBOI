```r
library(gridExtra)

# Monthly
monthly_prior = filter(busboi_monthly, ID2=='prior') %>% transmute(reach_id, prior_NSE=NSE)
monthly_algo  = filter(busboi_monthly, ID2=='algo')  %>% transmute(reach_id, algo_NSE=NSE)
monthly_diff  = left_join(monthly_prior, monthly_algo, by='reach_id') %>%
    mutate(delta_NSE = algo_NSE - prior_NSE) %>%
    filter(algo_NSE > -100)

percPos_m = 1 - (sum(monthly_diff$delta_NSE > 0) / length(monthly_diff$delta_NSE))

pm4 = ggplot(monthly_diff) +
    geom_point(aes(x=prior_NSE, y=algo_NSE), col=palette[3]) +
    geom_line(aes(x=prior_NSE, y=prior_NSE), col='black') +
    xlim(c(-2,1)) + ylim(c(-2,1)) +
    labs(x='Prior NSE', y='BUSBOI NSE', title='Monthly prior') +
    theme(text=element_text(size=16))

pm5 = ggplot(monthly_diff) +
    stat_ecdf(aes(x=delta_NSE), lwd=1, col=palette[3]) +
    coord_cartesian(xlim=c(-0.5,1)) +
    labs(x='NSE difference BUSBOI - Prior', y='CDF', title='Monthly prior') +
    annotate("text", hjust=0, x=0.03+round(quantile(monthly_diff$delta_NSE, 0.50), 2), y=0.52,
             label=round(quantile(monthly_diff$delta_NSE, 0.50), 2), size=5) +
    annotate("text", hjust=0, x=-0.5, y=0.52, label='median', size=5) +
    annotate("text", hjust=0, x=0.02, y=percPos_m-0.02, label=100*(1-round(percPos_m,2)), size=5) +
    annotate("text", hjust=0, x=0.5,  y=percPos_m-0.02, label='percent positive', size=5) +
    geom_hline(yintercept=0.50,     linetype='dotted') +
    geom_hline(yintercept=percPos_m, linetype='dotted') +
    theme(text=element_text(size=16))

pm6 = ggplot(monthly_diff) +
    geom_histogram(aes(x=delta_NSE), bins=6000, fill=palette[3]) +
    coord_cartesian(xlim=c(-2,2)) +
    labs(x='NSE difference BUSBOI - Prior', y='Count', title='Monthly prior') +
    theme(text=element_text(size=16))

# Daily
daily_prior = filter(busboi_daily, ID2=='prior') %>% transmute(reach_id, prior_NSE=NSE)
daily_algo  = filter(busboi_daily, ID2=='algo')  %>% transmute(reach_id, algo_NSE=NSE)
daily_diff  = left_join(daily_prior, daily_algo, by='reach_id') %>%
    mutate(delta_NSE = algo_NSE - prior_NSE) %>%
    filter(algo_NSE > -100)

percPos_d = 1 - (sum(daily_diff$delta_NSE > 0) / length(daily_diff$delta_NSE))

pd4 = ggplot(daily_diff) +
    geom_point(aes(x=prior_NSE, y=algo_NSE), col=palette[2]) +
    geom_line(aes(x=prior_NSE, y=prior_NSE), col='black') +
    xlim(c(-2,1)) + ylim(c(-2,1)) +
    labs(x='Prior NSE', y='BUSBOI NSE', title='Daily prior') +
    theme(text=element_text(size=16))

pd5 = ggplot(daily_diff) +
    stat_ecdf(aes(x=delta_NSE), lwd=1, col=palette[2]) +
    coord_cartesian(xlim=c(-0.5,1)) +
    labs(x='NSE difference BUSBOI - Prior', y='CDF', title='Daily prior') +
    annotate("text", hjust=0, x=0.03+round(quantile(daily_diff$delta_NSE, 0.50), 2), y=0.52,
             label=round(quantile(daily_diff$delta_NSE, 0.50), 2), size=5) +
    annotate("text", hjust=0, x=-0.5, y=0.52, label='median', size=5) +
    annotate("text", hjust=0, x=0.02, y=percPos_d-0.02, label=100*(1-round(percPos_d,2)), size=5) +
    annotate("text", hjust=0, x=0.5,  y=percPos_d-0.02, label='percent positive', size=5) +
    geom_hline(yintercept=0.50,     linetype='dotted') +
    geom_hline(yintercept=percPos_d, linetype='dotted') +
    theme(text=element_text(size=16))

pd6 = ggplot(daily_diff) +
    geom_histogram(aes(x=delta_NSE), bins=6000, fill=palette[2]) +
    coord_cartesian(xlim=c(-2,2)) +
    labs(x='NSE difference BUSBOI - Prior', y='Count', title='Daily prior') +
    theme(text=element_text(size=16))
```

```r
combined_diff = rbind(mutate(monthly_diff, ID='Monthly prior'),
                      mutate(daily_diff,   ID='Daily prior'))

p_scatter = ggplot(combined_diff) +
    geom_point(aes(x=prior_NSE, y=algo_NSE, col=ID), alpha=0.5) +
    geom_line(aes(x=prior_NSE, y=prior_NSE), col='black') +
    scale_color_manual(values=c('Monthly prior'=palette[3], 'Daily prior'=palette[2])) +
    xlim(c(-2,1)) + ylim(c(-2,1)) +
    labs(x='Prior NSE', y='BUSBOI NSE') +
    theme(text=element_text(size=16), legend.title=element_blank())

p_cdf = ggplot(combined_diff) +
    stat_ecdf(aes(x=delta_NSE, col=ID), lwd=1) +
    scale_color_manual(values=c('Monthly prior'=palette[3], 'Daily prior'=palette[2])) +
    coord_cartesian(xlim=c(-0.5,1)) +
    labs(x='NSE difference BUSBOI - Prior', y='CDF') +
    geom_hline(yintercept=0.50, linetype='dotted') +
    geom_hline(yintercept=percPos_m, linetype='dotted', col=palette[3]) +
    geom_hline(yintercept=percPos_d, linetype='dotted', col=palette[2]) +
    # median labels
    annotate("text", hjust=0, x=-0.5, y=0.52, label='median', size=5) +
    annotate("text", hjust=0, col=palette[3],
             x=0.03+round(quantile(monthly_diff$delta_NSE, 0.50), 2), y=0.52,
             label=paste0(round(quantile(monthly_diff$delta_NSE, 0.50), 2)), size=5) +
    annotate("text", hjust=0, col=palette[2],
             x=0.06, y=0.52,
             label=paste0(round(quantile(daily_diff$delta_NSE, 0.50), 2)), size=5) +
    # percent positive labels
    annotate("text", hjust=0, x=0.4, y=percPos_m+0.04, label='% positive', size=5) +
    annotate("text", hjust=0, col=palette[3],
             x=0.1, y=percPos_m+0.02,
             label=paste0(100*(1-round(percPos_m, 2)), '%'), size=5) +
    annotate("text", hjust=1, col=palette[2],
             x=-0.02, y=percPos_d+0.02,
             label=paste0(100*(1-round(percPos_d, 2)), '%'), size=5) +
    theme(text=element_text(size=16), legend.title=element_blank())

p_hist = ggplot(combined_diff) +
    geom_histogram(aes(x=delta_NSE, fill=ID), bins=6000, alpha=0.5, position='identity') +
    scale_fill_manual(values=c('Monthly prior'=palette[3], 'Daily prior'=palette[2])) +
    coord_cartesian(xlim=c(-2,2)) +
    labs(x='NSE difference BUSBOI - Prior', y='Count') +
    theme(text=element_text(size=16), legend.title=element_blank())

options(repr.plot.width=18, repr.plot.height=6)
combined = grid.arrange(p_scatter, p_cdf, p_hist, nrow=1)
ggsave('/nas/cee-water/cjgleason/colin/BUSBOI/BUSBOI/Figures/improvement2.png',
       combined, width=18, height=6, units='in', dpi=300)
```
